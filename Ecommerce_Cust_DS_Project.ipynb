```python
print("Duplicate Rows Before Cleaning:",df.duplicated().sum())
df=df.drop_duplicates().reset_index(drop=True)
print("Duplicate Rows After Cleaning:",df.duplicated().sum())
print("DataSet Shape After Removing DUplicates:",df.shape)
```

```python
df["Annual_Income"]=df["Annual_Income"].fillna(df["Annual_Income"].median())
df["Customer_Satisfaction"]=df["Customer_Satisfaction"].fillna(df["Customer_Satisfaction"].median())
df["Product_Rating"]=df["Product_Rating"].fillna(df["Product_Rating"].median())

#check misssing values after cleaning
print(df.isnull().sum())
print("Total Missing Values:",df.isnull().sum().sum())
```

```python
# Select numerical columns
numerical_columns = df.select_dtypes(include=np.number).columns

print("Numerical columns:")
print(numerical_columns.tolist())
```

```python
# Detect outliers using IQR method

outlier_summary = {}

for col in numerical_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = ((df[col] < lower_bound) | (df[col] > upper_bound)).sum()

    outlier_summary[col] = outliers

outlier_summary = pd.Series(outlier_summary).sort_values(ascending=False)

print(outlier_summary)
```

```python
# Outlier treatment using IQR capping

outlier_columns = [
    "Annual_Income",
    "Average_Order_Value",
    "Time_Spent_on_Website",
    "Customer_Tenure_Years",
    "Last_Purchase_Days_Ago",
    "Annual_Spend",
    "Customer_Satisfaction",
    "Product_Rating",
    "Number_of_Purchases",
    "Purchase_Frequency",
    "Cart_Abandonment_Rate"
]

for col in outlier_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    df[col] = df[col].clip(lower=lower_bound, upper=upper_bound)

print("Outlier treatment completed.")
```

```python
#Hypothesis Testing
#H₀ (Null): Loyalty and non-loyalty customers' average annual spend are not significantly different.
#H₁ (Alternative): Their average annual spend is significantly different.

from scipy.stats import ttest_ind

loyal_customers = df[df["Loyalty_Program"] == 1]["Annual_Spend"]
non_loyal_customers = df[df["Loyalty_Program"] == 0]["Annual_Spend"]

t_stat, p_value = ttest_ind(
    loyal_customers,
    non_loyal_customers,
    equal_var=False
)

print("T-statistic:", t_stat)
print("P-value:", p_value)
```

```python
alpha = 0.05

if p_value < alpha:
    print("Reject H0: The difference in annual spend is statistically significant.")
else:
    print("Fail to reject H0: No statistically significant difference was found.")

#Why this Matters?
#I performed hypothesis testing to determine whether loyalty-program membership was associated with a significant difference in customer spending.
```

```python
#Feature Engineering
print("Categorical Columns:")
print(df.select_dtypes(include="object").columns.tolist())

print("\nNumerical Columns:")
print(df.select_dtypes(include=np.number).columns.tolist())
```

```python
# One-Hot Encoding for categorical variables

categorical_cols = df.select_dtypes(include="object").columns.tolist()

df_encoded = pd.get_dummies(
    df,
    columns=categorical_cols,
    drop_first=True
)

print("Original shape:", df.shape)
print("Encoded shape:", df_encoded.shape)
```

```python
#ML Objective: customers will purchace or not ?
#SO Purchase_status is our target(y).

# Separate features and target

X = df_encoded.drop(columns=["Purchase_Status", "Customer_ID"])
y = df_encoded["Purchase_Status"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)

print("\nTarget distribution:")
print(y.value_counts())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

# Create model
logistic_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

# Train model
logistic_model.fit(X_train, y_train)

# Predictions
y_pred_logistic = logistic_model.predict(X_test)

# Accuracy
accuracy = accuracy_score(y_test, y_pred_logistic)

print("Logistic Regression Accuracy:", accuracy)

print("\nClassification Report:")
print(classification_report(y_test, y_pred_logistic))
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

# Scale numerical features
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Logistic Regression with scaled data
logistic_model_scaled = LogisticRegression(
    max_iter=1000,
    solver="lbfgs",
    random_state=42
)

logistic_model_scaled.fit(X_train_scaled, y_train)

# Predictions
y_pred_logistic_scaled = logistic_model_scaled.predict(X_test_scaled)

# Evaluation
print("Logistic Regression Accuracy:",
      accuracy_score(y_test, y_pred_logistic_scaled))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_logistic_scaled))
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

# Random Forest Model
rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

# Train
rf_model.fit(X_train, y_train)

# Prediction
y_pred_rf = rf_model.predict(X_test)

# Evaluation
print("Random Forest Accuracy:", accuracy_score(y_test, y_pred_rf))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_rf))
```

```python
# Model Comparison

model_results = {
    "Logistic Regression": accuracy_score(y_test, y_pred_logistic_scaled),
    "Random Forest": accuracy_score(y_test, y_pred_rf)
}

for model, accuracy in model_results.items():
    print(f"{model}: {accuracy:.4f}")
```

```python
from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

cm = confusion_matrix(y_test, y_pred_logistic_scaled)

plt.figure(figsize=(6, 5))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["No Purchase", "Purchase"],
    yticklabels=["No Purchase", "Purchase"]
)

plt.title("Confusion Matrix - Logistic Regression")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

```python
# Feature Importance - Logistic Regression

feature_importance = pd.DataFrame({
    "Feature": X_train.columns,
    "Coefficient": logistic_model_scaled.coef_[0]
})

feature_importance["Absolute_Coefficient"] = (
    feature_importance["Coefficient"].abs()
)

feature_importance = feature_importance.sort_values(
    "Absolute_Coefficient",
    ascending=False
)

print(feature_importance)
```
